```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('Sales_Clean_Data.csv')
```

Converting gender into numbers

```python
df_encoded = pd.get_dummies(data=df, columns=['gender'], drop_first=True)
```

Separating X from Y

```python
X  = df_encoded.drop(columns=['churn'])
y = df_encoded['churn']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

# TRAIMING LE MODELS

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

model = DecisionTreeClassifier(random_state=13)
model.fit(X_train, y_train)
model_predicts = model.predict(X_test)

print("Decision Tree Accuracy", accuracy_score(y_test, model_predicts))
```

Random forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV  #For tuning hyperparameters

model = RandomForestClassifier(random_state=13)
model.fit(X_train, y_train)
model_predicts = model.predict(X_test)

print("Random Forest Accuracy", accuracy_score(y_test, model_predicts))

# -------------------------------------------
# Fine tuning
# -------------------------------------------

param_grid = {
    'n_estimators' : [50, 100, 200],
    'max_depth' : [None, 10, 20],
    'min_samples_split' : [2, 5]
}

# Searching for the best combination
grid_search = GridSearchCV(RandomForestClassifier(random_state=13), param_grid, cv=5)
grid_search.fit(X_train, y_train)

tuned_model = grid_search.best_estimator_
print("Best parameters:", grid_search.best_params_)
```

Saving model to disk

```python
import joblib

joblib.dump(tuned_model, 'SalesML_Model.joblib')
joblib.dump(scaler, 'scaler.joblib')
```

K-Nearest Neighbor

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=13)
model.fit(X_train_scaled, y_train)
model_predicts = model.predict(X_test_scaled)

print("K-Nearest Neighbor Accuracy: ", accuracy_score(y_test, model_predicts))
```
